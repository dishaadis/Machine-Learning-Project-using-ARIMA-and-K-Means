```python
### Uang saku per hari
uang_saku_harian = 50000

# Total hari kuliah (Senin - Jumat)
jumlah_hari = 5

# Hitung total uang saku selama 5 hari
total_uang_saku = uang_saku_harian * jumlah_hari

# Input pengeluaran setiap hari (manual, tanpa for)
senin = int(input("Masukkan pengeluaran hari Senin: Rp "))
selasa = int(input("Masukkan pengeluaran hari Selasa: Rp "))
rabu = int(input("Masukkan pengeluaran hari Rabu: Rp "))
kamis = int(input("Masukkan pengeluaran hari Kamis: Rp "))
jumat = int(input("Masukkan pengeluaran hari Jumat: Rp "))

# Hitung total pengeluaran
total_pengeluaran = senin + selasa + rabu + kamis + jumat

# Hitung rata-rata pengeluaran per hari
rata_pengeluaran = total_pengeluaran / jumlah_hari

# Membuat laporan
laporan = (
    "Total uang saku selama 5 hari: Rp " + str(total_uang_saku) + "\n"
    + "Total pengeluaran: Rp " + str(total_pengeluaran) + "\n"
    + "Rata-rata pengeluaran per hari: Rp " + str(int(rata_pengeluaran))
)

print("\n=== Laporan Mingguan ===")
print(laporan)
```
